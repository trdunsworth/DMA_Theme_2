# DMA Theme Data-Visualization Palettes

Brewer-style schemes built from DMA brand colors.
Run this notebook from `palettes/python/` or add it to `sys.path`.

```python
import sys
sys.path.insert(0, ".")

import dma_palette as dma
import matplotlib_dma          # registers DMA_* colormaps
import matplotlib.pyplot as plt
import numpy as np
print("DMA palettes", dma.__version__)
```

## Qualitative (categorical)

```python
from matplotlib.colors import ListedColormap
fig, ax = plt.subplots(figsize=(8, 1.6))
ax.imshow(np.arange(8).reshape(1, -1), cmap=ListedColormap(dma.qualitative()))
ax.set_xticks(range(8)); ax.set_yticks([])
for i, c in enumerate(dma.qualitative()):
    ax.text(i, -0.7, c, ha="center", fontsize=8)
ax.set_title("DMA Qualitative (8)")
```

## Sequential ramps

```python
schemes = ["Blues", "Teals", "Turquoises", "Greens", "Oranges", "Cool", "Ocean", "Forest"]
fig, axes = plt.subplots(len(schemes), 1, figsize=(7, 0.9 * len(schemes)))
for ax, s in zip(axes, schemes):
    colors = dma.sequential(s, 9)
    ax.imshow(np.arange(9).reshape(1, -1), cmap=ListedColormap(colors))
    ax.set_yticks([]); ax.set_xticks([]); ax.set_ylabel(s, rotation=0,
        ha="right", va="center")
fig.suptitle("DMA Sequential (light -> dark)", y=1.02)
```

## Diverging

```python
divs = list(dma.DIVERGING_SCHEMES)
fig, axes = plt.subplots(len(divs), 1, figsize=(7, 1.1 * len(divs)))
for ax, s in zip(axes, divs):
    colors = dma.diverging(s, 11)
    ax.imshow(np.arange(11).reshape(1, -1), cmap=ListedColormap(colors))
    ax.set_yticks([]); ax.set_xticks([]); ax.set_ylabel(s.replace(' ', '\n'),
        rotation=0, ha="right", va="center", fontsize=8)
fig.suptitle("DMA Diverging", y=1.03)
```

## Matplotlib colormaps

After importing `matplotlib_dma`, every scheme is available by name (`"DMA Blues"`, `"DMA Cool"`, `"DMA Div Red-Blue"`).

```python
x = np.linspace(-3, 3, 300); y = np.linspace(-2, 2, 200)
X, Y = np.meshgrid(x, y); Z = np.exp(-(X**2 + Y**2)) + 0.5*np.sin(3*X)*np.cos(2*Y)

fig, axes = plt.subplots(1, 3, figsize=(13, 3.2), layout="constrained")
for ax, cm_name in zip(axes, ["DMA Blues", "DMA Cool", "DMA Div Red-Green"]):
    im = ax.contourf(X, Y, Z, levels=24, cmap=cm_name)
    fig.colorbar(im, ax=ax); ax.set_title(cm_name)
```

## Categorical cycle

```python
matplotlib_dma.use_dma_cycle()
fig, ax = plt.subplots(figsize=(6, 3))
t = np.linspace(0, 2*np.pi, 100)
for k in range(8):
    ax.plot(t, np.sin(t + k * np.pi / 5) + k * 0.4, label=f"series {k}")
ax.legend(ncol=4, fontsize=8); ax.set_title("Default cycle = DMA qualitative")
```

## Semantic status colors (WCAG AA on white)

```python
sem = dma.semantic()
fig, ax = plt.subplots(figsize=(6, 1.4))
ax.barh(range(4), [1]*4, color=list(sem.values()))
ax.set_yticks(range(4), [f"{k}  {v}" for k, v in sem.items()]); ax.set_xticks([])
ax.set_title("Status colors — all ≥ 4.5:1 contrast")
```
